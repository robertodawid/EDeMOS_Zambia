```python
import numpy as np
import pandas as pd
import zipfile
```

```python
zipfile_path = '../data/IHS_survey/MWI_2019_IHS-V_v06_M_CSV.zip'

data_folder = '../Data/IHS_survey/'


#open the zip file
with zipfile.ZipFile(zipfile_path, 'r') as z:
    #list the contents of the zip file
    #print(z.namelist())
    
    with z.open('HH_MOD_E.csv') as f:  #Time Use and Labour
        #read the csv 
        df1 = pd.read_csv(f)
        
        print(df1.head(5))

    # with z.open('householdgeovariables_ihs5.csv') as f: #geographic data
    #     #read the csv 
    #     df2 = pd.read_csv(f)
        
    #     print(df2.head(5))

    with z.open('HH_MOD_B.csv') as f: #sex
        #read the csv 
        df3 = pd.read_csv(f)
        
        print(df3.head(5))
    
    with z.open('hh_mod_a_filt.csv') as f: #geographic data  urban/rural
        #read the csv 
        df4 = pd.read_csv(f)
        
        print(df4.head(5))
   
    # with z.open('HH_MOD_B.csv') as f: #household members
    #     #read the csv 
    #     df5 = pd.read_csv(f)
        
    #     print(df5.head(5))
```

```python
df4
```

```python
# hh_b03        sex
# hh_e06_8a    economic activity did [NAME]spent most time in the last 12 months (ACT 1)

# main supposition: is that rows for df 1 and 3 are in the same order, so we can merge them on the index
df = pd.concat([df1[['case_id', 'PID', 'hh_e06_8a']], df3[['hh_b03']]], axis=1)
df = pd.pivot_table(df, index=['case_id', 'hh_b03'], columns=['hh_e06_8a'], aggfunc='size', fill_value=0).reset_index()
df = pd.merge(df, df4[['case_id', 'district', 'reside']], on='case_id', how='left')
df
```

### Filter the households used in the PCA

```python
hh_id = pd.read_csv('../data/IHS_survey/household_data_IHS_case_id.csv')
hh_id
```

```python
df = df[df['case_id'].isin(hh_id['case_id'])]
df
```

```python
# Step 1: Group by district and gender
gender_counts = df.groupby(['district', 'reside', 'hh_b03']).size().unstack(fill_value=0)

# Step 2: Calculate share of men and women
gender_share = gender_counts.div(gender_counts.sum(axis=1), axis=0)

# Step 3: Rename for clarity
gender_share = gender_share.rename(columns={
    'FEMALE': 'Share_Female',
    'MALE': 'Share_Male'
})

gender_share = gender_share.reset_index().set_index('district')
gender_share['reside'] = gender_share['reside'].str.lower()
# Final result
gender_share
```

```python
occupation_share_men = gender_share[['reside', 'Share_Male']].copy()
occupation_share_women = gender_share[['reside', 'Share_Female']].copy()
```

```python
# Save the two DataFrame to a CSV file
occupation_share_men.to_csv(data_folder + 'employee_survey_men.csv')
occupation_share_women.to_csv(data_folder + 'employee_survey_women.csv')
```
